```python
# ==============================================================================
# Caso de Uso Real: Comparativa de Ventas
# ==============================================================================
import pandas as pd

print("--- 1. Entorno Preparado ---")
print("Librería Pandas importada con éxito.\n")

# ------------------------------------------------------------------------------
# CELDA 1: Creación de Series de Ventas (Listas y Diccionarios)
# ------------------------------------------------------------------------------
# Ventas del Trimestre 1 (Enero - Marzo) y Trimestre 2 (Abril - Junio) por Mes
meses_t1 = ['Enero', 'Febrero', 'Marzo']
ventas_mensuales_t1 = pd.Series([12500, 15300, 11800], index=meses_t1)

meses_t2 = ['Abril', 'Mayo', 'Junio']
ventas_mensuales_t2 = pd.Series([18200, 14000, 19500], index=meses_t2)

# Ventas por Sucursal para ambos trimestres
ventas_t1 = pd.Series({
    'Sucursal Norte': 45000,
    'Sucursal Sur': 32000,
    'Sucursal Centro': 58000,
    'Sucursal Este': 29000,
    'Sucursal Oeste': 41000
})

ventas_t2 = pd.Series({
    'Sucursal Norte': 48000,
    'Sucursal Sur': 35000,
    'Sucursal Centro': 62000,
    'Sucursal Este': 31000,
    'Sucursal Oeste': 39000
})

print("--- 1. Estructuras Creadas ---")
print("Ventas T1 por Sucursal:\n", ventas_t1)
print("\nVentas T2 por Sucursal:\n", ventas_t2)
print("\n" + "="*60 + "\n")


print("--- 2. Inspección Metadatólica (Propiedades) ---")
print(f"Número de sucursales evaluadas (.size): {ventas_t1.size}")
print(f"Estructura del Índice (.index): {ventas_t1.index}")
print(f"Tipo de Datos Almacenados (.dtype): {ventas_t1.dtype}")
print("\n" + "="*60 + "\n")


print("--- 3. Indexación y Slicing ---")
# Rebanado posicional: Primeros 2 meses del T1 (Enero y Febrero)
print("Ventas de los primeros 2 meses de T1 (Slicing 0:2):")
print(ventas_mensuales_t1[0:2])

# Consulta explícita por etiqueta
print(f"\nIngresos de la 'Sucursal Centro' en T1: ${ventas_t1['Sucursal Centro']:,}")

# Selección múltiple por lista de etiquetas
print("\nComparativa puntual Norte vs Sur en T1:")
print(ventas_t1[['Sucursal Norte', 'Sucursal Sur']])
print("\n" + "="*60 + "\n")

print("--- 4. Indicadores Estadísticos Agregados (T1) ---")
print(f"Ingresos Totales (T1): ${ventas_t1.sum():,}")
print(f"Sucursal Máxima (T1): ${ventas_t1.max():,}")
print(f"Sucursal Mínima (T1): ${ventas_t1.min():,}")
print(f"Desviación Estándar de Ventas (T1): ${ventas_t1.std():,.2f}")

print("\nResumen Descriptivo Completo de Ventas T1:")
print(ventas_t1.describe())
print("\n" + "="*60 + "\n")

meta_ingresos = 40000
sucursales_top_t1 = ventas_t1[ventas_t1 > meta_ingresos]

print(f"--- 5. Filtrado Condicional (Sucursales T1 > ${meta_ingresos:,}) ---")
print(sucursales_top_t1)
print("\n" + "="*60 + "\n")

print("--- 6. Reorganización y Ordenamiento (T1) ---")
print("Ranking por Ingresos (Valores Ascendentes):")
print(ventas_t1.sort_values())

print("\nSucursales Ordenadas Alfabéticamente (Índice Descendente Z-A):")
print(ventas_t1.sort_index(ascending=False))
print("\n" + "="*60 + "\n")

print("--- 7. Asignación Escalar y Operaciones Vectorizadas (T1 vs T2) ---")

# Presupuesto operativo base asignado a todas las sucursales mediante un valor escalar
presupuesto_base = pd.Series(5000, index=ventas_t1.index)
print("Presupuesto Operativo Asignado por Sucursal (Valor Escalar):\n", presupuesto_base)

# Operaciones Vectorizadas entre Series: T2 - T1
crecimiento_absoluto = ventas_t2 - ventas_t1
crecimiento_porcentual = ((ventas_t2 - ventas_t1) / ventas_t1) * 100
ventas_semestre_1 = ventas_t1 + ventas_t2

print("\n--- COMPARATIVA SEMESTRAL CONSOLIDADA ---")
print("Crecimiento Neto Absoluto ($ T2 - T1):\n", crecimiento_absoluto)
print("\nCrecimiento Porcentual (% T2 vs T1):\n", crecimiento_porcentual.round(2).astype(str) + '%')
print("\nTotal Acumulado Semestre 1 (T1 + T2):\n", ventas_semestre_1)

# Sucursal con mejor crecimiento relativo
top_crecimiento = crecimiento_porcentual.idxmax()
print(f"\n-> La sucursal con mayor crecimiento en T2 fue '{top_crecimiento}' con un +{crecimiento_porcentual.max():.2f}%")
```
