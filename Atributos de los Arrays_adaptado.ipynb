```python
import numpy as np

# Caso adaptado: ventas de dos trimestres
ventas_trimestre_1 = np.array([120, 150, 180])
ventas_trimestre_2 = np.array([130, 170, 200])

# Suma de ventas
total_ventas = ventas_trimestre_1 + ventas_trimestre_2

print("Ventas trimestre 1:", ventas_trimestre_1)
print("Ventas trimestre 2:", ventas_trimestre_2)
print("Total de ventas:", total_ventas)

# Atributos de los arrays
print("Shape:", total_ventas.shape)
print("Size:", total_ventas.size)
print("Dimensiones:", total_ventas.ndim)
print("Tipo de dato:", total_ventas.dtype)
```
