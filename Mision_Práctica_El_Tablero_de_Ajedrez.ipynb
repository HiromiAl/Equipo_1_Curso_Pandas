# Misión Práctica Equipo 1: Tablero de Ajedrez 8x8 con NumPy y Slicing
En esta sección construimos un tablero de ajedrez (matriz 8x8 de 0s y 1s) utilizando la librería NumPy, inspeccionamos sus atributos principales y extraemos subsecciones mediante la técnica de slicing.


```python
import numpy as np
```

#### 1. Creación del tablero 8x8 lleno de ceros

```python
tablero = np.zeros((8, 8), dtype=int)
```

#### 2. Asignación del patrón de ajedrez (unos intercalados)



```python
# Filas pares (0, 2, 4, 6), columnas impares (1, 3, 5, 7)
tablero[::2, 1::2] = 1

# Filas impares (1, 3, 5, 7), columnas pares (0, 2, 4, 6)
tablero[1::2, ::2] = 1

print("=== TABLERO DE AJEDREZ (8x8) ===")
print(tablero)
```

Visualización

```python

import matplotlib.pyplot as plt

# Visualización gráfica del tablero
plt.figure(figsize=(4, 4))
plt.imshow(tablero, cmap='gray')
plt.title("Visualización del Tablero (8x8)", fontsize=11, fontweight='bold')
plt.xticks(range(8))
plt.yticks(range(8))
plt.show()
```

### 3. Inspección de Atributos del Array

```python
print("\n=== ATRIBUTOS DEL TABLERO ===")
print("Forma (.shape):", tablero.shape)
print("Dimensiones (.ndim):", tablero.ndim)
print("Total de elementos (.size):", tablero.size)
print("Tipo de dato (.dtype):", tablero.dtype)
```

###4. Extracción de Secciones Específicas usando Slicing

```python
cuadrante_sup_izq = tablero[0:4, 0:4]   # Primeras 4 filas y 4 columnas
centro_tablero = tablero[2:6, 2:6]      # Submatriz central de 4x4
esquina_inf_der = tablero[6:8, 6:8]     # Submatriz de 2x2 en la esquina inferior

print("\n=== EXTRACCIÓN CON SLICING (SUBSECCIONES) ===")
print("1. Cuadrante Superior Izquierdo (4x4):\n", cuadrante_sup_izq)
print("\n2. Centro del Tablero (4x4):\n", centro_tablero)
print("\n3. Esquina Inferior Derecha (2x2):\n", esquina_inf_der)
```

Visualización

```python
import matplotlib.patches as patches
import matplotlib.pyplot as plt

# Configurar 4 subgráficos en una sola fila
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

# 1. Tablero con las zonas de Slicing marcadas (Primer subgráfico: axes[0])
axes[0].imshow(tablero, cmap='gray')
axes[0].set_title('1. Zonas de Slicing', fontsize=10, fontweight='bold')

# Marcos de colores
marco_rojo = patches.Rectangle(
    (-0.5, -0.5), 4, 4, linewidth=3, edgecolor='red', facecolor='none'
)
marco_azul = patches.Rectangle(
    (1.5, 1.5), 4, 4, linewidth=3, edgecolor='deepskyblue', facecolor='none'
)
marco_verde = patches.Rectangle(
    (5.5, 5.5), 2, 2, linewidth=3, edgecolor='lime', facecolor='none'
)

axes[0].add_patch(marco_rojo)
axes[0].add_patch(marco_azul)
axes[0].add_patch(marco_verde)

# 2. Cuadrante Superior Izquierdo (Rojo) -> axes[1]
axes[1].imshow(cuadrante_sup_izq, cmap='gray')
axes[1].set_title('Sup. Izquierdo [0:4, 0:4]', fontsize=10)
for spine in axes[1].spines.values():
  spine.set_edgecolor('red')
  spine.set_linewidth(3)

# 3. Centro del Tablero (Azul) -> axes[2]
axes[2].imshow(centro_tablero, cmap='gray')
axes[2].set_title('Centro [2:6, 2:6]', fontsize=10)
for spine in axes[2].spines.values():
  spine.set_edgecolor('deepskyblue')
  spine.set_linewidth(3)

# 4. Esquina Inferior Derecha (Verde) -> axes[3]
axes[3].imshow(esquina_inf_der, cmap='gray')
axes[3].set_title('Esquina Inf. Der [6:8, 6:8]', fontsize=10)
for spine in axes[3].spines.values():
  spine.set_edgecolor('lime')
  spine.set_linewidth(3)

plt.tight_layout()
plt.show()
```
